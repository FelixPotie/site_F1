# TP26 - Épreuve 1 : Le temple 

!!! info Règle
Vous êtes arrivé.e sur l'île de l'informatique pour trouver le trésor de M.Potié, avancez d'épreuve en épreuve jusqu'au butin !

!!!

!!! abstract 
Une fois arrivé sur l'île vous marchez jusqu'à un temple, vous y rentrez mais la porte se ferme. \
Vous y trouvez une carte vous montrant où est caché le trésor. Il faut maintenant sortir de ce temple par la porte qui mène au bon endroit. Cependant 3 portes mènent à la sortie mais une seule permet vraiment de sortir.

Pour sortir vous devrez trouver la bonne porte et déchiffrer le code d'accès (code capytale du prochain TP).
!!!
!!! question 
Une inscription est écrite :\
**Vous décrypterez le message en ajoutant 8 à tout le monde ! La bonne porte est celle qui donne la complexité de la recherche dichotomique.**

- Sur la porte de droite est écrit $O(n log(n))$ et le code est `1Y1^%))*.+--(`
- Sur la porte central est écrit $O(n)$ et le code est : `11+)%))*.+-*1`
- Sur la porte de gauche est écrit $O(log(n))$ et le code est : `1+\\-%))*.+*,*`

!!!
!!! example 
Rappel : on peut récupéré le numéro ascii d'un caractère `c` en faisant `ord(c)` et le caractère depuis un entier `k` en faisant `chr(k)`.
!!!
