# TP24 - Tri rapide (Quicksort)

Nous avons vu de nombreux algorithmes de tri. Voilà le dernier algorithme au programme à voir.


!!! info Principe
L'algorithme quicksort est un algorithme de type **Diviser pour régner**. Un élément pivot va être utilisé pour récursivement diviser le tableau.

- Trouver l'élément pivot (plusieurs façons de le faire, ici ça sera le dernier élément du tableau)
- Créer deux nouveaux tableaux, puis itérer sur le tableau à trier : tous les éléments plus petits que le pivot se retrouvent dans le 1er tableau, tous les éléments plus grands au pivot dans le deuxième tableau. Le pivot doit être exclu de ces comparaisons
- Concaténer le tableau des plus petits éléments (à trier par récursion) avec le pivot puis avec le tableau des plus grands éléments (à trier par récursion)
!!!

!!! abstract Pour comparer les temps d'exécution :
Pour ça on va utiliser le module `time`. 
```python

```
Pour savoir le temps d'exécution d'un algorithme on peut faire la différence entre le temps d'exécution avant l'algo et après.
!!!


!!! question Question 5
Tracer avec Matplotlib des courbes de temps d'executions de vos 3 algorithmes en fonction de la taille de votre liste d'entrée.
!!!

````python
import matplotlib.pyplot as plt
import time
import random

def affiche_comparaison_tri(max_valeur=10000, max_taille=1000):
    algos = [
        # PARTIE À MODIFIER AVEC LES NOMS DE FONCTIONS QUE VOUS AVEZ CODÉ
        (tri_rapide_rand, "Tri Rapide", "#FF5733"),
        (tri_fusion, "Tri Fusion", "#33FF57"), 
        (tri_selection, "Tri Sélection", "#3357FF"),
        (tri_insertion, "Tri Insertion", "#F333FF"),
        (tri_bulle, "Tri Bulle", "#FFBD33"),
        (tri_comptage, "Tri Comptage", "#33FFF5")
    ]

    x = []
    resultats = {nom: [] for _, nom, _ in algos}

    for n in range(0, max_taille, 10): 
        x.append(n)
        liste_originale = [random.randrange(max_valeur) for _ in range(n)]

        for fonction_tri, nom, _ in algos:
            liste_test = liste_originale.copy()
            
            temps_debut = time.perf_counter()
            fonction_tri(liste_test)
            temps_fin = time.perf_counter()
            
            resultats[nom].append(temps_fin - temps_debut)

    plt.figure(figsize=(10, 6))
    plt.title("Comparaison des performances des algorithmes de tri")
    plt.xlabel("Nombre d'éléments dans la liste")
    plt.ylabel("Temps d'exécution (secondes)")

    for _, nom, couleur in algos:
        plt.plot(x, resultats[nom], label=nom, color=couleur, linewidth=2)

    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()
````

````python
affiche_comparaison_tri(max_valeur=10000, max_taille=500)
````
