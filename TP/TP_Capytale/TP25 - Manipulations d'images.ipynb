!!! abstract Représentation
Une image numérique en couleur est représentée par un tableau à 3 dimensions :
- La **hauteur** (nombre de lignes de pixels)
- La **largeur** (nombre de colonnes de pixels)
- Les **canaux de couleur** : 
    - En couleur : 3 valeurs entre 0 et 255 pour le rouge (R), le vert (G) et le bleu (B)
    - Nuance de gris : 1 valeurs entre 0 et 255
!!!

!!! abstract Fonctions utiles
On va utiliser les bibliothèques numpy, PIL (module Image) et matplotlib module pyplot.\

Voici les principales fonctions et attributs que nous allons utiliser :

| Fonction / Attribut | Description |
|---|---|
| `Image.open("nom_img.png")` | Créé une image PIL depuis une source |
| `np.array(img_pil)` | Convertit une image PIL en tableau NumPy |
| `Image.fromarray(tableau)` | Convertit un tableau NumPy en image PIL |
| `tableau.shape` | Renvoie les dimensions  `(hauteur, largeur, canaux)` |
| `np.zeros((h, w, 3), dtype=np.uint8)` | Crée un tableau noir de dimensions h×w |
| `plt.imshow(tableau)` | Affiche un tableau sous forme d'image |
!!!

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
```

```python
img_pil = Image.open("thones.png")
image = np.array(img_pil)

plt.figure()
plt.imshow(image)
plt.axis("off")
plt.show()
```

## 3. Modification par convolution

!!! abstract Qu'est-ce que la convolution ?
La **convolution** est une opération mathématique fondamentale en traitement d'images. Elle consiste à faire glisser un petit tableau appelé **noyau** (ou *kernel*) sur l'image et à calculer, pour chaque pixel, la **somme pondérée** de ses voisins.\

Ici nous allons nous restreindre à un noyau 3x3. Donc on va faire la moyenne des 9 pixels dont le pixels qu'on regarde est au centre.
!!!
